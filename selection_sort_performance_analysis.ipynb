```python
import random
import time

def selection_sort(list):
    comparisons = 0
    swaps = 0
    for i in range(len(list)):
        min_index = i
        for j in range(i + 1, len(list)):
            comparisons += 1
            if list[j] < list[min_index]:
                min_index = j
        if min_index != i:
            list[i], list[min_index] = list[min_index], list[i]
            swaps += 1
    return list, comparisons, swaps

print(selection_sort([3, 7, 2, 4, 1])[0])
print()
sizes = [103, 104]
for n in sizes:
    random_list = random.sample(range(n * 10), n)
    scenarios = {
        "Random": random_list,
        "Sorted": sorted(random_list),
        "Reversed": sorted(random_list, reverse=True),
    }
    print("n =", n)
    for name, data in scenarios.items():
        start = time.time()
        result, comparisons, swaps = selection_sort(data.copy())
        elapsed = time.time() - start
        print(f"  {name:<9} Comparisons: {comparisons:<10} Swaps: {swaps:<7} Time: {elapsed:.4f} s")
    print()
```
